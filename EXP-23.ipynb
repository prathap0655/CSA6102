```
# ==========================================================
# Experiment 6: Bit Stream Copy
# Aim:
# Create a bit stream copy of a simulated storage device and
# prove using SHA-256 hashing that the copy is a forensically
# exact replica of the original.
# ==========================================================

import os
import hashlib

# ----------------------------------------------------------
# Step 1: Create Original Evidence File
# ----------------------------------------------------------
original_path = "original_evidence.bin"

# Simulate a storage device (1024 bytes)
with open(original_path, "wb") as file:
    file.write(os.urandom(1024))

# ----------------------------------------------------------
# Step 2: Create Bit Stream Copy
# ----------------------------------------------------------
def bit_stream_copy(src_path, dst_path):
    with open(src_path, "rb") as src:
        with open(dst_path, "wb") as dst:
            dst.write(src.read())

copy_path = "bitstream_copy.bin"
bit_stream_copy(original_path, copy_path)

# ----------------------------------------------------------
# Step 3: Calculate SHA-256 Hash
# ----------------------------------------------------------
def sha256_of_file(path):
    sha256 = hashlib.sha256()

    with open(path, "rb") as file:
        while True:
            chunk = file.read(4096)
            if not chunk:
                break
            sha256.update(chunk)

    return sha256.hexdigest()

original_hash = sha256_of_file(original_path)
copy_hash = sha256_of_file(copy_path)

# ----------------------------------------------------------
# Step 4: Display Output
# ----------------------------------------------------------
print("Original hash:", original_hash)
print("Copy hash:    ", copy_hash)

# ----------------------------------------------------------
# Step 5: Test Cases
# ----------------------------------------------------------
def test_experiment6():

    assert os.path.getsize(original_path) == os.path.getsize(copy_path), \
        "Bit stream copy must match original size exactly"

    assert original_hash == copy_hash, \
        "Hashes must match, proving the copy is forensically identical"

    print("Experiment 6: All test cases passed.")

test_experiment6()
```
