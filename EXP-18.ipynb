```
import os
import hashlib

evidence_dir = "digital_evidence"
os.makedirs(evidence_dir, exist_ok=True)

sample_evidence = {
    "email_log.txt": """From: attacker@mail.com
To: victim@mail.com
Subject: Invoice
Please pay immediately.""",

    "browser_history.txt": """http://malicious-site.com/login
http://bank.com/transfer""",

    "chat_message.txt": """Hey, did you send the file?""",

    "system_log.txt": """2026-07-24 10:15:32 - USB device connected: E:"""
}

for filename, content in sample_evidence.items():
    filepath = os.path.join(evidence_dir, filename)

    with open(filepath, "w") as file:
        file.write(content)

print("Sample evidence files created successfully.\n")


def catalog_evidence(folder):
    catalog = []

    for filename in sorted(os.listdir(folder)):
        filepath = os.path.join(folder, filename)

        file_size = os.stat(filepath).st_size

        with open(filepath, "rb") as file:
            data = file.read()

        md5_hash = hashlib.md5(data).hexdigest()

        catalog.append({
            "filename": filename,
            "size_bytes": file_size,
            "md5": md5_hash
        })

    return catalog


evidence_catalog = catalog_evidence(evidence_dir)

print("DIGITAL EVIDENCE CATALOG")
print("-" * 60)

for item in evidence_catalog:
    print(f"Filename   : {item['filename']}")
    print(f"Size(Bytes): {item['size_bytes']}")
    print(f"MD5 Hash   : {item['md5']}")
    print("-" * 60)


def test_experiment1():

    assert os.path.isdir(evidence_dir), \
        "Evidence directory should exist"

    assert len(evidence_catalog) == len(sample_evidence), \
        "All evidence files should be catalogued"

    filenames = [entry["filename"] for entry in evidence_catalog]

    for expected_file in sample_evidence:
        assert expected_file in filenames, \
            f"{expected_file} missing from catalog"

    for entry in evidence_catalog:
        assert entry["size_bytes"] > 0, \
            "File size should be greater than zero"

        assert len(entry["md5"]) == 32, \
            "Invalid MD5 hash"

    print("\nExperiment 1: All test cases passed successfully.")


test_experiment1()
```
