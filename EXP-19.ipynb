```
# Experiment 2: Volatile vs Non-Volatile Evidence Capture

import os
import sys
import subprocess

# Install psutil if not available
try:
    import psutil
except ImportError:
    subprocess.check_call(
        [sys.executable, "-m", "pip", "install", "psutil"]
    )
    import psutil

# Create evidence directory
evidence_dir = "digital_evidence"
os.makedirs(evidence_dir, exist_ok=True)

# Create sample evidence files
sample_evidence = {
    "email_log.txt": """From: attacker@mail.com
To: victim@mail.com
Subject: Invoice
Please pay immediately.""",

    "browser_history.txt": """http://malicious-site.com/login
http://bank.com/transfer""",

    "chat_message.txt": "Hey, did you send the file?",

    "system_log.txt": "2026-07-24 10:15:32 - USB device connected: E:"
}

# Create files
for filename, content in sample_evidence.items():

    filepath = os.path.join(evidence_dir, filename)

    with open(filepath, "w") as file:
        file.write(content)

print("Sample evidence files created successfully.\n")


# Capture volatile evidence
def capture_volatile_evidence():

    processes = [
        p.info
        for p in psutil.process_iter(["pid", "name"])
    ]

    return {
        "running_process_count": len(processes),
        "sample_processes": processes[:5]
    }


# Capture non-volatile evidence
def capture_nonvolatile_evidence(folder):

    files = sorted(os.listdir(folder))

    return {
        "disk_files": files,
        "file_count": len(files)
    }


# Capture snapshots
volatile_snapshot = capture_volatile_evidence()

nonvolatile_snapshot = capture_nonvolatile_evidence(
    evidence_dir
)


# Print output
print(
    "Volatile evidence -> running processes:",
    volatile_snapshot["running_process_count"]
)

print(
    "Non-volatile evidence -> disk files:",
    nonvolatile_snapshot
)


# Test cases
def test_experiment2():

    assert volatile_snapshot["running_process_count"] > 0, \
        "There should be at least one running process"

    assert isinstance(
        volatile_snapshot["sample_processes"],
        list
    )

    assert nonvolatile_snapshot["file_count"] == len(
        sample_evidence
    )

    snapshot2 = capture_nonvolatile_evidence(
        evidence_dir
    )

    assert snapshot2 == nonvolatile_snapshot, \
        "Non-volatile evidence must remain unchanged between reads"

    print("Experiment 2: All test cases passed.")


# Run tests
test_experiment2()
```
