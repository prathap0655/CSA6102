```
# ==========================================================
# Experiment 9: File System Feature Simulation
# Aim:
# Model simplified file entries for FAT, NTFS and EXT to
# demonstrate that FAT lacks permissions and journaling,
# while NTFS and EXT support both.
# ==========================================================

# ----------------------------------------------------------
# Step 1: Define FAT File Entry
# ----------------------------------------------------------
class FATFileEntry:
    def __init__(self, name, size):
        self.name = name
        self.size = size
        # FAT does not support permissions or journaling

# ----------------------------------------------------------
# Step 2: Define NTFS File Entry
# ----------------------------------------------------------
class NTFSFileEntry:
    def __init__(self, name, size, permissions="rw-r--r--"):
        self.name = name
        self.size = size
        self.permissions = permissions
        self.journal_entry = f"WRITE {name} {size} bytes"

# ----------------------------------------------------------
# Step 3: Define EXT File Entry
# ----------------------------------------------------------
class EXTFileEntry:
    def __init__(self, name, size, permissions="rw-r--r--"):
        self.name = name
        self.size = size
        self.permissions = permissions
        self.journal_entry = f"WRITE {name} {size} bytes"

# ----------------------------------------------------------
# Step 4: Create Sample File Entries
# ----------------------------------------------------------
fat_file = FATFileEntry("data.txt", 1024)
ntfs_file = NTFSFileEntry("data.txt", 1024)
ext_file = EXTFileEntry("data.txt", 1024)

# ----------------------------------------------------------
# Step 5: Display Output
# ----------------------------------------------------------
print("FAT has permissions attribute:", hasattr(fat_file, "permissions"))
print("NTFS has permissions attribute:", hasattr(ntfs_file, "permissions"))
print("EXT has permissions attribute:", hasattr(ext_file, "permissions"))

# ----------------------------------------------------------
# Step 6: Test Cases
# ----------------------------------------------------------
def test_experiment9():

    assert not hasattr(fat_file, "permissions"), \
        "FAT should not support file permissions"

    assert not hasattr(fat_file, "journal_entry"), \
        "FAT should not support journaling"

    assert hasattr(ntfs_file, "permissions") and \
           hasattr(ntfs_file, "journal_entry"), \
        "NTFS should support permissions and journaling"

    assert hasattr(ext_file, "permissions") and \
           hasattr(ext_file, "journal_entry"), \
        "EXT should support permissions and journaling"

    print("Experiment 9: All test cases passed.")

# ----------------------------------------------------------
# Step 7: Run Test Cases
# ----------------------------------------------------------
test_experiment9()
```
